```python
import numpy as np
file_list = [1, 2, 3]
filename_prefix = "results/results_japan_averaged_rate"
cost_zr = np.zeros((4, 3), dtype=float)
cost_zrPlus = np.zeros((4, 3), dtype=float)
cost_trans = np.zeros((4, 3), float)
cost_ip = np.zeros((4, 3), dtype=float)
```

```python
for j in range(3):
    #for file in file_list:
    file = file_list[j]
    filename = filename_prefix + str(file) + "_1.txt"
    with open(filename, "r") as f:
        lines = f.readlines()
        for i in range(4):
            # line = lines[1 + 2 * (i+1)]
            line = lines[2 * (i + 1)].split()
            cost_zr[i][j] = line[0]
            cost_zrPlus[i][j] = line[1]
            cost_trans[i][j] = line[3]
            cost_ip[i][j] = line[2]
print(cost_zr)
print(cost_zrPlus)
print(cost_trans)
print(cost_ip)

cost_zrStr = []
cost_zrPlusStr = []
cost_transStr = []
cost_ipStr = []
for i in range(4):
    cost_zrStr.append([0, 0, 0])
    cost_zrPlusStr.append([0, 0, 0])
    cost_transStr.append([0, 0, 0])
    cost_ipStr.append([0, 0, 0])

for i in range(4):
    for j in range(3):
        cost_zrStr[i][j] = str(int(cost_zr[i][j]))
        cost_zrPlusStr[i][j] = str(int(cost_zrPlus[i][j]))
        cost_transStr[i][j] = str(int(cost_trans[i][j]))
        cost_ipStr[i][j] = str(int(cost_ip[i][j]))

with open("resultsPaper/@resultsCost.csv", "a") as f:
    for i in range(4):
        line = ", ".join(cost_zrStr[i])
        line = line + " , " + ", ".join(cost_zrPlusStr[i])
        line = line + " , " + ", ".join(cost_transStr[i])
        line = line + " , " + ", ".join(cost_ipStr[i])
        line = line + "\n"
        f.write(line)
```
